```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
merged = pd.read_csv("kolkata_aqi_merged_final.csv")

print("Final Dataset Shape:", merged.shape)
print("\nColumns:", merged.columns.tolist())
print("\nFirst 3 rows preview:")
print(merged.head(3))
```

```python
print("Final Dataset Shape:", merged.shape)
print("\nColumns:", merged.columns.tolist())
print("\nFirst 3 rows preview:")
print(merged.head(3))
```

```python
plt.figure(figsize=(15, 10))
```

```python
plt.subplot(2, 2, 1)
sns.histplot(merged['AQI'], kde=True, bins=30)
plt.title('AQI Distribution Kolkata')
```

```python
plt.subplot(2, 2, 2)
sns.scatterplot(x=merged['temperature_2m'], y=merged['AQI'])
plt.title('AQI vs Temperature')
```

```python
plt.subplot(2, 2, 3)
sns.boxplot(x='hour', y='AQI', data=merged)
plt.title('AQI by Hour of Day')
```

```python
plt.subplot(2, 2, 4)
corr = merged[['AQI', 'temperature_2m', 'relative_humidity_2m', 'wind_speed_10m', 
               'rain', 'hour', 'is_weekend']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Feature Correlation')

plt.tight_layout()
plt.show()
```

```python
corr.to_csv("correlation_matrix.csv")
```
